Dataset Summary Report Using Python and Pandas
1. Introduction

This notebook demonstrates how to load and explore a student dataset using Python and Pandas. The analysis focuses on understanding the basic structure, characteristics, data types, and missing values present in the dataset.

The dataset contains information about students, including their ID, name, age, course, marks, attendance percentage, and city.

2. Import Pandas

```python
import pandas as pd
```

3. Load the Dataset

```python
df = pd.read_csv("Day7_Student_Dataset.csv")
print("Dataset loaded successfully!")
```

5. Display the Last Few Rows

13. Check Missing Values

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

Missing Values Found

The dataset contains missing values in:

Marks → 1 missing value
Attendance_Percentage → 1 missing value
City → 1 missing value

The remaining columns do not contain missing values.

14. Calculate Total Missing Values

```python
total_missing = df.isnull().sum().sum()

print("Total Missing Values:", total_missing)
```

15. Check Unique Values

```python
for column in df.columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

16. Basic Dataset Summary

```python
print("===== DATASET SUMMARY =====")
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nFeatures:")
for column in df.columns:
  print("-", column)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```
